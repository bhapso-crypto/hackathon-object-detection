```python
import os
import kagglehub

# 1. Download dataset (read-only cache location)
dataset_dir = kagglehub.dataset_download(
    "mugheesahmad/sh17-dataset-for-ppe-detection"
)

# 2. Save data.yaml in writable /content/ directory
yaml_path = "/content/data.yaml"

# 3. Reference dataset_dir in the 'path' key
yaml_content = f"""path: {dataset_dir}
train: images
val: images

nc: 17
names:
  0: person
  1: ear
  2: earmuffs
  3: face
  4: face-guard
  5: face-mask-medical
  6: foot
  7: tools
  8: glasses
  9: gloves
  10: helmet
  11: hands
  12: head
  13: medical-suit
  14: shoes
  15: safety-suit
  16: safety-vest
"""

with open(yaml_path, "w") as f:
  f.write(yaml_content)

print("✅ data.yaml created successfully at:", yaml_path)
```

```python
from ultralytics import YOLO

model = YOLO("yolo26n.pt")

results = model.train(
    data="/content/data.yaml",  # Direct path to writable config file
    project="/content/SH17_outputs",
    name="sh17_model-26n-200",
    epochs=200,
    imgsz=640,
    batch=32,
    device=0,
    cache=True,
    workers=8,
    amp=True,
    save=True,  # Guarantees saving best.pt and last.pt
    save_period=50,  # Saves a checkpoint after every single epoch
    val=True,  # Runs validation every epoch to evaluate accuracy
    verbose=True,  # Displays detailed per-class stats table
)
```
